```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.cluster import KMeans
```

```python
data = pd.read_csv("mall_customer.csv")
```

```python
numeric_features = [
    'Age','IncomeLevel','ElectronicsSpending','ClothingSpending',
    'GrocerySpending','HomeSpending','Visits','PurchaseFrequency',
    'OnlineActivity','EmailOpens','AppUsage','LoyaltyPoints'
]

categorical_features = ['Gender', 'MembershipLevel']
```

```python
scaler = StandardScaler()
numeric_scaled = scaler.fit_transform(data[numeric_features])
```

```python
encoder = OneHotEncoder(sparse_output=False)
categorical_encoded = encoder.fit_transform(data[categorical_features])
```

```python
X = np.hstack([numeric_scaled, categorical_encoded])
```

```python
for k in [2, 3, 4]:
    print("\nK =", k)
    kmeans = KMeans(n_clusters=k, random_state=42)
    clusters = kmeans.fit_predict(X)
    data['Cluster'] = clusters

    print("\nNumber of customers per cluster:")
    print(data['Cluster'].value_counts())

    print("\nAverage values per cluster:")
    print(
        data.groupby('Cluster')[
            [
                'ElectronicsSpending',
                'ClothingSpending',
                'GrocerySpending',
                'HomeSpending',
                'IncomeLevel',
                'LoyaltyPoints'
            ]
        ].mean()
    )
```
